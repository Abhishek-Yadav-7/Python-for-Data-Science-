5b. Write a python code to show statistical information on given data set.

```python
import statistics

print(statistics.mean([1, 3, 5, 7, 9, 11, 13]))
print(statistics.median([1, 3, 5, 7, 9, 11]))
print(statistics.mode([1, 1, 3, -5, 7, -9, 11]))
print(statistics.stdev([2, 2.5, 1.25, 3.1, 1.75, 2.8]))
print(statistics.variance([2, 2.5, 1.25, 3.1, 1.75, 2.8]))


print("\nAbhishek_46")
```

5c.Write a python code to create pandas series from dictionaries. 


```python

import pandas as pd

dictionary = {'Abhishek': 10, 'Pradeep': 20, 'Divyanshu': 30}


series = pd.Series(dictionary)

print(series)

print("\nAbhishek_46")
```

```python

import pandas as pd

dictionary = {'Abhishek': 50, 'Pradeep': 10, 'Divyanshu': 80}

series = pd.Series(dictionary, index=['Pradeep', 'Divyanshu', 'D', 'Abhishek'])

print(series)

print("\nAbhishek_46")
```

5d. Write a python code to demonstrate filter pandas series with Boolean arrays.

```python

import pandas as pd
 

dict = {'name':["Abhishek", "Pradeep", "Divyanshu", "Geeku"],
        'degree': ["MBA", "BCA", "M.Tech", "MBA"],
        'score':[90, 40, 80, 98]}
 
df = pd.DataFrame(dict, index = [True, False, True, False])
 
print(df)

print("\nAbhishek_46")
```
